```python
import pandas as pd
import numpy as np
```

```python
tips = pd.read_csv("https://www.dropbox.com/s/sidhaoro98iej0j/tips.csv?raw=1",index_col=0)
#tips = pd.read_csv("tips.csv",index_col=0)
```

```python
tips.head()
```

```python
#不同日期该餐厅顾客总人数和每条账单的平均人数
tips.pivot_table(values='size',index='day',aggfunc = {np.mean,np.sum})
```

```python
#不同时间内男顾客和女顾客的人数
tips.pivot_table(values='size', index=['time'], columns='sex',aggfunc = sum,margins=True)
```

```python
#不同性别之间有无吸烟者的账单条数
tips["count"] = 1
tips.pivot_table(values='count', index=['smoker'], columns='sex',aggfunc = sum,margins=True)
```

```python
#该餐厅更受哪类顾客喜爱
tips["satisfication"] = tips["tip"]/tips["total_bill"]
tips.pivot_tab·le(values='satisfication',index=['smoker','sex'])
```
