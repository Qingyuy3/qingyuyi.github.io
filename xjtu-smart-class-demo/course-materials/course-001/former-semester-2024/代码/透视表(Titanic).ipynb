```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("titanic.csv")
```

```python
df
```

```python
df.info()
```

```python
df.drop(axis=1,labels=['Cabin'],inplace=True)
```

```python
df.dropna(axis=0,how="any",inplace=True)
```

```python
df.info()
```

```python
#不同等级船舱男乘客与女乘客的生存率和生存人数
df.pivot_table(values='Survived',index="Sex",columns="Pclass",aggfunc={np.mean,np.sum})
```

```python
#不同等级船舱乘客平均年龄
df.pivot_table(values='Age',index=["Pclass"],columns="Survived")
```

```python
#统计乘客兄弟姐妹个数对于生存率的影响
df.pivot_table(index="SibSp", values="Survived")
```

```python
#统计乘客父母小孩个数对于生存率的影响
df.pivot_table(index="Parch", values="Survived")
```

```python
#不同等级船舱“18岁以下”“18-45岁”“45岁以上”生存人数
bins = [min(df['Age'])-1, 18, 45, max(df['Age'])]
bins_label = ['18岁及以下', '18-45岁', '45岁以上']
df['Age_cut'] = pd.cut(df['Age'], bins=bins, labels=bins_label)
```

```python
df.pivot_table(index="Age_cut",values="Survived",columns="Pclass",aggfunc=np.sum)
```
